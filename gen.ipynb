```python
import numpy as np
import pandas as pd
```

```python
# Definir las letras y coordenadas iniciales
# Coordenadas básicas para las letras "S", "I", "T", "E", "M" en X y Y
letters = {
    'S': [(-11, 11), (-10, 11), (-9, 11), (-9, 12), (-9, 13), (-10, 13), (-11, 13), (-11, 14), (-11, 15), (-10, 15), (-9, 15)],
    'I': [(-7, 15), (-6, 15), (-5, 15), (-6, 15), (-6, 14), (-6, 13), (-6, 12), (-6, 11), (-7, 11), (-6, 11), (-5, 11)],
    'T': [(-1, 11), (-1, 12), (-1, 13), (-1, 14), (-1, 15), (-2, 15), (-3, 15), (-2, 15), (-1, 15), (0, 15), (1, 15)],
    'E': [(3, 15), (4, 15), (5, 15), (4, 15), (3, 15), (3, 14), (3, 13), (4, 13), (5, 13), (4, 13), (3, 13), (3, 12), (3, 11), (4, 11), (5, 11)],
    'M': [(7, 11), (7, 12), (7, 13), (7, 14), (7, 15), (8, 14), (9, 13), (10, 14), (11, 15), (11, 14), (11, 13), (11, 12), (11,11)],
}
```

```python
# Espacio de trabajo en MATLAB
xmin_matlab = 0.40
xmax_matlab = 0.55
ymin_matlab = 2.45
ymax_matlab = 2.48

# Encuentra los rangos actuales de tus coordenadas
all_x = [x for letter in letters.values() for (x,y) in letter]
all_y = [y for letter in letters.values() for (x,y) in letter]

xmin_py = min(all_x)
xmax_py = max(all_x)
ymin_py = min(all_y)
ymax_py = max(all_y)

# Función para escalar coordenadas
def scale_coord(x, y):
    x_scaled = xmin_matlab + (x - xmin_py) * (xmax_matlab - xmin_matlab) / (xmax_py - xmin_py)
    y_scaled = ymin_matlab + (y - ymin_py) * (ymax_matlab - ymin_matlab) / (ymax_py - ymin_py)
    return (x_scaled, y_scaled)

# Aplicar escalamiento a todas las letras
letters_scaled = {}
for letter, points in letters.items():
    letters_scaled[letter] = [scale_coord(x, y) for (x,y) in points]
```

```python
# Combinar todos los valores en una sola lista
all_positions = []
for lista in letters_scaled.values():
    all_positions.extend(lista)
```

```python
# Convertir a DataFrame
all_positions = np.array(all_positions)
df = pd.DataFrame({
    'X': all_positions[:, 0],
    'Y': all_positions[:, 1],
    'Z': 0  # Z siempre es 0
})

```

```python
# Exportar a CSV
output_path = "./SITEM_positions.csv"
df.to_csv(output_path, index=False, header=False)
output_path
```
